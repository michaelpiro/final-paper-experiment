# Camera-ready performance tables (Pavia + San Diego I/II)
Downloads the `camera-ready-v1` release checkpoints, computes the
train-background scores for the CFAR thresholds, and emits the updated
Pavia table (core rows = the exact printed Table 1) + the two San Diego
tables with the same criteria. ~15-25 min on a T4 (deep + GMM-Levin
train scorings dominate).


```python
!git clone -b tsp-repro --depth 1 https://github.com/michaelpiro/final-paper-experiment.git repo
%cd repo
import sys, torch
sys.path.insert(0, '.')
import tsp_repro
from tsp_repro.iid_camera_ready import apply_robust_lrao
apply_robust_lrao()
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print('device:', DEVICE)
```

```python
import os, urllib.request, zipfile
URL = ('https://github.com/michaelpiro/final-paper-experiment/releases/'
       'download/camera-ready-v1/camera_ready_ckpts.zip')
if not os.path.exists('camera_ready_ckpts'):
    urllib.request.urlretrieve(URL, 'ckpts.zip')
    with zipfile.ZipFile('ckpts.zip') as z: z.extractall('camera_ready_ckpts')
print(sorted(os.listdir('camera_ready_ckpts')))
```

```python
from tsp_repro.camera_ready_tables import make_all
# checkpoint-only: raises (never trains) if any checkpoint is missing
make_all('camera_ready_ckpts', out_dir='tables_out', device=DEVICE)
```

```python
import zipfile, os
with zipfile.ZipFile('camera_ready_tables.zip', 'w') as z:
    for f in os.listdir('tables_out'): z.write(os.path.join('tables_out', f))
try:
    from google.colab import files; files.download('camera_ready_tables.zip')
except Exception: print('(not on Colab)')
```
